```python
import pandas as pd
import numpy as np
data = pd.DataFrame({
    'Color': ['Red', 'Red', 'Blue', 'Green'],
    'Size': ['Small', 'Medium', 'Large', 'Medium'],
    'Purchased': [1, 1, 0, 1]
})

def nominal_dissimilarity(data, columns):
    n = len(data)
    matrix = np.zeros((n, n))

    for i in range(n):
        for j in range(n):
            mismatches = sum(data.loc[i, col] != data.loc[j, col]
                             for col in columns)
            matrix[i, j] = mismatches / len(columns)

    return matrix

def ordinal_dissimilarity(data, column, order):
    ranks = {value: i for i, value in enumerate(order)}
    normalized = data[column].map(ranks) / (len(order) - 1)

    n = len(data)
    matrix = np.zeros((n, n))

    for i in range(n):
        for j in range(n):
            matrix[i, j] = abs(normalized.iloc[i] - normalized.iloc[j])

    return matrix
def binary_dissimilarity(data, column):
    n = len(data)
    matrix = np.zeros((n, n))

    for i in range(n):
        for j in range(n):
            a = data.loc[i, column]
            b = data.loc[j, column]

            if a == 1 and b == 1:
                matrix[i, j] = 0
            elif a != b:
                matrix[i, j] = 1
            else:
                matrix[i, j] = 0

    return matrix
nominal_matrix = nominal_dissimilarity(data, ['Color'])

ordinal_matrix = ordinal_dissimilarity(
    data,
    'Size',
    ['Small', 'Medium', 'Large']
)

binary_matrix = binary_dissimilarity(data, 'Purchased')

print("Nominal Dissimilarity Matrix:")
print(nominal_matrix)

print("\nOrdinal Dissimilarity Matrix:")
print(ordinal_matrix)

print("\nBinary Dissimilarity Matrix:")
print(binary_matrix)
```

```python
import pandas as pd
import numpy as np

data =pd.DataFrame({
    'student' : [1,1,0,0],
    'Employed' : [0,1,1,0],
    'purchased': [1,1,0,1]
})

print("BINARY DATASET")
print(data)

n = len(data)
matrix = np.zeros((n,n))

for i in range(n):
  for j in range(n):

    diff = 0

    for col in data.columns:
      if data.loc[i,col] != data.loc[j,col]:
        diff += 1

    matrix[i,j] = diff / len(data.columns)
print("\nDISSIMILARITY MATRIX - BINARY")
print(pd.DataFrame(
    matrix,
    index = ['01','02','03','04'],
    columns = ['01','02','03','04']
).round(2)
)
```

```python
import pandas as pd
import numpy as np

data =pd.DataFrame({
    'satisfaction': ['Low','medium','high','medium']
})

print("ORDINAL DATASET")
print(data)

rank ={
    'low': 0,
    'medium': 1,
    'high': 2
}

n = len(data)
matrix = np.zeros((n,n))

for i in range(n):
  for j in range(n):

    x = rank[data.loc[i,'satisfaction'].lower()]
    y = rank[data.loc[j,'satisfaction'].lower()]

    matrix[i,j] = abs(x-y)/2

print("\nDISSIMILARITY MATRIX - ORDINAL")
print(pd.DataFrame(
    matrix,
    index = ['01','02','03','04'],
    columns = ['01','02','03','04']
).round(2))
```
